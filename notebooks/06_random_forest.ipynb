```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
X_train = pd.read_csv("../data/final/X_train.csv")
X_test = pd.read_csv("../data/final/X_test.csv")

y_train = pd.read_csv("../data/final/y_train.csv")
y_test = pd.read_csv("../data/final/y_test.csv")
```

```python
y_train = y_train.squeeze()
y_test = y_test.squeeze()
```

```python
rf_model = RandomForestClassifier(
    random_state=42
)
```

```python
rf_model.fit(
    X_train,
    y_train
)

print("Random Forest trained!")
```

```python
y_pred_rf = rf_model.predict(X_test)

y_prob_rf = rf_model.predict_proba(X_test)[:,1]
```

```python
print(
    classification_report(
        y_test,
        y_pred_rf
    )
)
```

```python
roc_auc_rf = roc_auc_score(
    y_test,
    y_prob_rf
)

print(
    f"ROC-AUC: {roc_auc_rf:.4f}"
)
```

```python
rf_balanced = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)
```

```python
rf_balanced.fit(
    X_train,
    y_train
)

print("Balanced Random Forest trained!")
```

```python
y_pred_rf_balanced = rf_balanced.predict(X_test)

y_prob_rf_balanced = rf_balanced.predict_proba(X_test)[:,1]
```

```python
print(
    classification_report(
        y_test,
        y_pred_rf_balanced
    )
)
```

```python
roc_auc_rf_balanced = roc_auc_score(
    y_test,
    y_prob_rf_balanced
)

print(
    f"ROC-AUC: {roc_auc_rf_balanced:.4f}"
)
```

```python
import joblib

joblib.dump(
    rf_balanced,
    "../models/random_forest_balanced.pkl"
)

print("Balanced Random Forest saved!")
```
