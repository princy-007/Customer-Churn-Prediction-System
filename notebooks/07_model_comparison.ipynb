```python
import pandas as pd
import numpy as np

import joblib

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
X_test = pd.read_csv("../data/final/X_test.csv")

y_test = pd.read_csv("../data/final/y_test.csv")

y_test = y_test.squeeze()

print(X_test.shape)
print(y_test.shape)
```

```python
logistic_model = joblib.load(
    "../models/logistic_regression.pkl"
)

decision_tree_model = joblib.load(
    "../models/decision_tree_final.pkl"
)

random_forest_model = joblib.load(
    "../models/random_forest_balanced.pkl"
)

print("All models loaded successfully!")
```

```python
# Logistic Regression predictions

y_pred_lr = logistic_model.predict(X_test)
y_prob_lr = logistic_model.predict_proba(X_test)[:,1]


# Decision Tree predictions

y_pred_dt = decision_tree_model.predict(X_test)
y_prob_dt = decision_tree_model.predict_proba(X_test)[:,1]


# Random Forest predictions

y_pred_rf = random_forest_model.predict(X_test)
y_prob_rf = random_forest_model.predict_proba(X_test)[:,1]


print("Predictions generated successfully!")
```

```python
model_comparison = pd.DataFrame({

    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Balanced Random Forest"
    ],

    "Accuracy": [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf)
    ],

    "Precision": [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_dt),
        precision_score(y_test, y_pred_rf)
    ],

    "Recall": [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_dt),
        recall_score(y_test, y_pred_rf)
    ],

    "F1 Score": [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_dt),
        f1_score(y_test, y_pred_rf)
    ],

    "ROC-AUC": [
        roc_auc_score(y_test, y_prob_lr),
        roc_auc_score(y_test, y_prob_dt),
        roc_auc_score(y_test, y_prob_rf)
    ]
})


model_comparison
```

```python
joblib.dump(
    logistic_model,
    "../models/final_churn_model.pkl"
)

print("Final churn model saved successfully!")
```
